#1.NumPy

```python
import numpy as np

numbers = np.array([1, 2, 3, 4])

print(numbers)
```

```python
print(numbers.dtype)
print(numbers.shape)
print(numbers.ndim)
print(numbers.size)
print(numbers*5)
```

### Index

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[-1])
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr[0, 1])
```

### Slicing

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
```

### Mathematical operations

```python
arr = np.array([10, 20, 30, 40])

print(arr + 5)
print(arr - 5)
print(arr * 2)
print(arr / 2)
```

### NumPy Funtions

```python
arr = np.zeros(5)

print(arr)
```

```python
arr = np.ones(5)

print(arr)
```

```python
arr = np.arange(1, 10)

print(arr)
```

```python
arr = np.linspace(0, 3, 5)

print(arr)
```

```python
arr = np.random.rand(5)

print(arr)
```

```python
arr = np.random.randint(1, 10, 5)

print(arr)
```

### Mean,Median,Maximum,Minimum

```python
marks = np.array([70, 80, 90, 60, 100])
print(np.mean(marks))
print(np.median(marks))
print(np.std(marks))
print(np.max(marks))
print(np.min(marks))
print(np.sum(marks))
```

### Flatten

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

new_arr = arr.flatten()

print(new_arr)
```

### Axis (0,1)

```python
data = np.array([
    [20, 50000, 2],
    [30, 60000, 5],
    [40, 80000, 10]
])
```

```python
mean_values = data.mean(axis=0)

print(mean_values)
```

```python
mean_values = data.mean(axis=1)

print(mean_values)
```

```python
predictions = np.array([
    [0.1, 0.7, 0.2],
    [0.8, 0.1, 0.1],
    [0.2, 0.2, 0.6]
])

classes = np.argmax(predictions, axis=1)

print(classes)
```

### Normalization

```python


data = np.array([0, 50, 100, 150, 200, 250])

normalized = data / 250

print(normalized)
```

### Standardization

```python


data = np.array([10, 20, 30, 40, 50])

mean = np.mean(data)
std = np.std(data)

standardized = (data - mean) / std

print(standardized)
```

### seed random

```python
np.random.seed(42)

data = np.random.randint(1, 101, 10)

print(data)
```

```python
rng = np.random.default_rng(42)

numbers = rng.random(5)

print(numbers)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.concatenate((a, b))

print(result)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.vstack((a, b))

print(result)
```

```python
age = np.array([25, 30, 35])
salary = np.array([40000, 50000, 60000])

g = np.stack((age, salary), axis=1)

print(g)
```

```python
data = np.array([10, 20, np.nan, 40, 50])

data[np.isnan(data)] = 0

print(data)
```

```python
data = np.array([10, 20, np.nan, 40, 50])

mean_value = np.nanmean(data)

data[np.isnan(data)] = mean_value

print(data)
```

```python
data = np.array([
    [10, 20, 30],
    [40, np.nan, 60],
    [70, 80, 90]
])

print(np.isnan(data))
```
