```python
from google.colab import userdata

api_key = userdata.get("GROQ_API_KEY")
print("API key loaded successfully")

from langchain_groq import ChatGroq

MODEL_NAME = "openai/gpt-oss-20b"  # llama-3.1-8b-instant was decommissioned by Groq on Aug 16, 2026

llm = ChatGroq(
    api_key=api_key,
    model=MODEL_NAME,
    temperature=0
)

def simple_chat(user_query):
    response = llm.invoke(user_query)
    return response.content

response1 = simple_chat("I want to learn AI")
print(response1)

print('____'*45)

response2 = simple_chat("How months its takes ?")
print(response2)
```
